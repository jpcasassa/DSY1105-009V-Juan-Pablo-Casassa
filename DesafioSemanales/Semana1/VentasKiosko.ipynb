```kotlin
var stock = 30
val recreos = 6
val bebidasPorRecreo = 4
val precio = 1200

var dinero = 0
var vendidas = 0

for (recreo in 1..recreos){
    if (stock >= recreos){
        stock = stock - bebidasPorRecreo
        vendidas = vendidas + bebidasPorRecreo
        dinero = dinero + (bebidasPorRecreo * precio)
    }
}
```
